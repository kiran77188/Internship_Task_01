```python
# Investigate negative quantities and cancellation invoices

cancellation_records = df.loc[
    (df["quantity"] < 0) &
    (df["invoiceno"].astype(str).str.startswith("C"))
]

display(cancellation_records.head(10))
```

```python
# Investigate cancelled transactions

cancellation_mask = df["invoiceno"].astype(str).str.startswith("C")

cancellation_records = df.loc[cancellation_mask]

print(f"Total cancellation records: {len(cancellation_records):}")
```

```python
# Identify negative quantities in cancellation and non-cancellation invoices

negative_quantity_mask = df["quantity"] < 0

cancellation_negative = df.loc[
    cancellation_mask & negative_quantity_mask
]

non_cancellation_negative = df.loc[
    ~cancellation_mask & negative_quantity_mask
]

print(f"Negative quantities in cancellation invoices: {len(cancellation_negative):}")
print(f"Negative quantities in other invoices: {len(non_cancellation_negative):}")
```

```python
# Analyze negative quantity records by product description

return_summary = (
    cancellation_negative
    .groupby("description", dropna=False)
    .agg(
        transaction_count=("quantity", "size"),
        total_quantity=("quantity", "sum")
    )
    .sort_values("transaction_count", ascending=False)
)

display(return_summary.head(10))
```

```python
# 8. Classify transactions based on business rules

cancellation_mask = (
    df["invoiceno"].astype(str).str.startswith("C")
)

bad_debt_mask = (
    df["description"].str.contains(
        "adjust bad debt",
        case=False,
        na=False
    )
)

df["transaction_type"] = "Normal / Other"

df.loc[
    cancellation_mask,
    "transaction_type"
] = "Cancellation"

df.loc[
    bad_debt_mask,
    "transaction_type"
] = "Accounting Adjustment"

print(df["transaction_type"].value_counts())
```

```python
# 10. Create separate analytical datasets

sales_df = df.loc[
    (df["transaction_type"] == "Normal / Other") &
    (df["quantity"] > 0) &
    (df["unitprice"] > 0)
].copy()

cancellation_df = df.loc[
    df["transaction_type"] == "Cancellation"
].copy()

adjustment_df = df.loc[
    df["transaction_type"] == "Accounting Adjustment"
].copy()
```

```python
# 1. Identify negative quantity and unit price records

negative_quantity = df["quantity"] < 0
negative_unitprice = df["unitprice"] < 0

print(f"Negative Quantity: {negative_quantity.sum():}")
print(f"Negative UnitPrice: {negative_unitprice.sum():}")
print(
    f"Both Negative: "
    f"{(negative_quantity & negative_unitprice).sum():}"
)
```

```python
# 2. Investigate negative quantity transactions

negative_quantity_records = df.loc[
    df["quantity"] < 0
].copy()

# Identify cancellation and non-cancellation invoices
negative_quantity_records["invoice_type"] = (
    negative_quantity_records["invoiceno"]
    .astype(str)
    .str.startswith("C")
    .map({True: "Cancellation", False: "Other"})
)

# Summarize negative quantity transactions
quantity_summary = (
    negative_quantity_records
    .groupby("invoice_type")
    .agg(
        record_count=("quantity", "size"),
        total_quantity=("quantity", "sum")
    )
)

display(quantity_summary)
```
